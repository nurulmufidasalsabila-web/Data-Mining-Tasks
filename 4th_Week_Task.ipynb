**PCA Visualization (Principal Component Analysis)**

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

features = df[[
    "danceability",
    "energy",
    "loudness",
    "speechiness",
    "acousticness",
    "instrumentalness",
    "liveness",
    "valence",
    "tempo"
]]

# normalisasi
scaler = StandardScaler()
scaled = scaler.fit_transform(features)

# PCA
pca = PCA(n_components=2)
pca_result = pca.fit_transform(scaled)

df["PCA1"] = pca_result[:,0]
df["PCA2"] = pca_result[:,1]

sns.scatterplot(x="PCA1", y="PCA2", hue="playlist_genre", data=df)

plt.title("PCA Visualization of Spotify Songs")
plt.show()
```

**K-Means Clustering Visualization**

```python
from sklearn.cluster import KMeans

X = scaled

kmeans = KMeans(n_clusters=5)
clusters = kmeans.fit_predict(X)

df["cluster"] = clusters

sns.scatterplot(
    x=df["danceability"],
    y=df["energy"],
    hue=df["cluster"],
    palette="Set2"
)

plt.title("K-Means Clustering of Songs")
plt.show()
```

**Hierarchical Clustering (Dendrogram)**

```python
from scipy.cluster.hierarchy import dendrogram, linkage

sample = scaled[:500]

linked = linkage(sample, method="ward")

plt.figure(figsize=(10,6))
dendrogram(linked)

plt.title("Hierarchical Clustering Dendrogram")
plt.show()
```

**t-SNE Visualization**

```python
from sklearn.manifold import TSNE

tsne = TSNE(n_components=2, random_state=42)

tsne_result = tsne.fit_transform(scaled[:2000])

plt.figure(figsize=(8,6))

plt.scatter(tsne_result[:,0], tsne_result[:,1])

plt.title("t-SNE Visualization of Songs")

plt.show()
```

**Radar Chart (Spider Chart)**

```python
import numpy as np

# fitur yang dibandingkan
features = ["danceability","energy","valence","acousticness","tempo"]

# ambil rata-rata tiap genre
genre_data = df.groupby("playlist_genre")[features].mean()

# pilih beberapa genre saja supaya grafik jelas
genres = genre_data.index[:5]

values = genre_data.loc[genres]

labels = features
num_vars = len(labels)

angles = np.linspace(0, 2*np.pi, num_vars, endpoint=False)
angles = np.concatenate((angles,[angles[0]]))

plt.figure(figsize=(6,6))

for genre in genres:

    data = values.loc[genre].values
    data = np.concatenate((data,[data[0]]))

    plt.polar(angles, data, label=genre)

plt.xticks(angles[:-1], labels)

plt.title("Radar Chart of Audio Features by Genre")

plt.legend(loc="upper right", bbox_to_anchor=(1.3,1.1))

plt.show()
```

```python
import networkx as nx

# ambil fitur numerik
numeric_df = df.select_dtypes(include=np.number)

# hitung korelasi
corr = numeric_df.corr()

# buat graph
G = nx.Graph()

# tambah node
for col in corr.columns:
    G.add_node(col)

# tambah edge jika korelasi cukup besar
for i in corr.columns:
    for j in corr.columns:
        if i != j and abs(corr.loc[i,j]) > 0.5:
            G.add_edge(i,j, weight=corr.loc[i,j])

plt.figure(figsize=(10,8))

pos = nx.spring_layout(G)

nx.draw(G, pos,
        with_labels=True,
        node_size=2000,
        node_color="lightblue",
        font_size=10)

plt.title("Network Graph of Audio Feature Correlations")

plt.show()
```
